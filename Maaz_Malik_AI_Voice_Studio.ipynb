```python
#@title 🔐 Step 1 — Malik Voice Studio Access Check { display-mode: "form" }
EMAIL = '' #@param {type:"string"}

# ==========================================================
# 1. SETTINGS (Yahan apni details likhein)
# ==========================================================
CONFIG = {
    "tool_name": "Malik Voice Studio",
    "owner_name": "Maaz",
    "admin_emails": ["maaz.malik0684@gmail.com"],
    "api_url": "YAHAN_APPS_SCRIPT_URL_PASTE_KAREIN",
    "secret_key": "malik_voice_secret_2026",
    "support_email": "maaz.malik0684@gmail.com",
    "whatsapp_number": "YAHAN_NUMBER_LIKHEIN",
    "whatsapp_link": "https://wa.me/923000000000",
    "wa_channel_name": "YAHAN_CHANNEL_NAME_LIKHEIN",
    "wa_channel_link": "https://whatsapp.com/channel/YAHAN_LINK",
    "social_links": {
        "YouTube": "",
        "Instagram": "",
        "TikTok": "",
    },
}

# ==========================================================
# 2. ACCESS VERIFICATION ENGINE (Isay chherne ki zaroorat nahi)
# ==========================================================
import urllib.request, urllib.parse, json, secrets, builtins
from IPython.display import display, HTML

builtins._ok = False
builtins._malik_session = None
builtins._malik_active_user = None

def _show_card(title, color, message):
    btns = []
    if CONFIG["whatsapp_link"] and "000000000" not in CONFIG["whatsapp_link"]:
        btns.append(f"<a href='{CONFIG['whatsapp_link']}' target='_blank' style='background:#25d366;color:#fff;padding:9px 16px;border-radius:8px;text-decoration:none;font-weight:700;margin-right:8px;display:inline-block;margin-top:10px;'>💬 WhatsApp</a>")
    if CONFIG["wa_channel_link"] and "YAHAN_" not in CONFIG["wa_channel_link"]:
        btns.append(f"<a href='{CONFIG['wa_channel_link']}' target='_blank' style='background:#128c7e;color:#fff;padding:9px 16px;border-radius:8px;text-decoration:none;font-weight:700;margin-right:8px;display:inline-block;margin-top:10px;'>📢 {CONFIG['wa_channel_name']}</a>")
    for name, url in CONFIG["social_links"].items():
        if url.strip():
            btns.append(f"<a href='{url.strip()}' target='_blank' style='background:#2b2538;color:#fff;border:1px solid #554b6e;padding:8px 14px;border-radius:8px;text-decoration:none;font-size:.88em;margin-right:8px;display:inline-block;margin-top:10px;'>🔗 {name}</a>")

    contact = []
    if CONFIG["support_email"]:
        contact.append(f"📧 <b>Email:</b> {CONFIG['support_email']}")
    if CONFIG["whatsapp_number"] and "YAHAN_" not in CONFIG["whatsapp_number"]:
        contact.append(f"💬 <b>WhatsApp:</b> {CONFIG['whatsapp_number']}")

    display(HTML(
        f"<div style='font-family:Inter,Arial,sans-serif;max-width:580px;background:#181320;border:1.5px solid {color};border-radius:14px;padding:22px 26px;color:#eee;'>"
        f"<div style='font-size:1.25em;font-weight:800;color:{color};'>{title}</div>"
        f"<div style='margin-top:10px;line-height:1.65;font-size:.95em;'>{message}</div>"
        f"<div style='margin-top:12px;line-height:1.8;font-size:.92em;'>{'<br>'.join(contact)}</div>"
        f"<div>{''.join(btns)}</div>"
        f"<div style='margin-top:14px;font-size:.8em;color:#998;'>{CONFIG['tool_name']} • By {CONFIG['owner_name']}</div>"
        f"</div>"
    ))
    raise SystemExit("Access stopped.")

em = EMAIL.strip().lower()
if not em:
    _show_card("⚠️ Email Required", "#f5b041", "Upar <b>EMAIL</b> box mein apni Gmail likhein aur dobara Run karein.")

if "YAHAN_" in CONFIG["api_url"]:
    _show_card("⚙️ Setup Incomplete", "#f5b041", "CONFIG mein apna <b>api_url</b> paste karein.")

# Step A: Asli Google account check
try:
    from google.colab import auth
    auth.authenticate_user()
    import google.auth, google.auth.transport.requests
    creds, _ = google.auth.default()
    creds.refresh(google.auth.transport.requests.Request())
    req = urllib.request.Request(f"https://www.googleapis.com/oauth2/v1/userinfo?access_token={creds.token}")
    colab_email = json.loads(urllib.request.urlopen(req, timeout=10).read().decode()).get("email", "").lower()
except Exception:
    colab_email = ""

if not colab_email:
    _show_card("⚠️ Login Failed", "#ff6b6b", "Google account verify nahi ho saka. Dobara sign in karke try karein.")

is_admin = colab_email in [a.lower() for a in CONFIG["admin_emails"]]
print(f"🔍 Google Account: {colab_email}")

if (not is_admin) and (colab_email != em):
    _show_card("🚫 Account Mismatch", "#ff6b6b", f"Aapka login account (<b>{colab_email}</b>) aur likhi hui email (<b>{em}</b>) mukhtalif hain. Apni asli Gmail use karein.")

# Step B: Sheet se Start, Expire aur Status check
try:
    url = f"{CONFIG['api_url']}?key={urllib.parse.quote(CONFIG['secret_key'])}&email={urllib.parse.quote(em)}"
    data = json.loads(urllib.request.urlopen(url, timeout=15).read().decode())
except Exception:
    _show_card("🌐 Connection Error", "#f5b041", "Server se rabta nahi ho saka. Internet check karke dobara try karein.")

status  = data.get("status", "")
plan    = data.get("plan", "Standard")
start_t = data.get("start", "")
exp_t   = data.get("expires", "")

if status == "ok":
    builtins._ok = True
    builtins._malik_session = secrets.token_hex(24)
    builtins._malik_active_user = {"email": em, "plan": plan, "tool_name": CONFIG["tool_name"]}
    print(f"✅ Access Granted! | Plan: {plan}")
    if start_t: print(f"   🟢 Start:  {start_t}")
    if exp_t:   print(f"   🔴 Expire: {exp_t}")
    print("👉 Ab neeche Step 2 run karein.")

elif status == "not_started":
    _show_card("🕒 Plan Not Started Yet", "#f5b041", f"Aapka plan (<b>{plan}</b>) abhi shuru nahi hua.<br><b>Start Time:</b> {start_t}<br><b>Expire Time:</b> {exp_t}")

elif status == "expired":
    _show_card("⏳ Subscription Expired", "#ff6b6b", f"Aapka <b>{CONFIG['tool_name']}</b> ka waqt khatam ho gaya hai.<br><b>Expired On:</b> {exp_t}<br>Renew karwane ke liye neeche rabta karein:")

elif status == "blocked":
    _show_card("🛑 Account Blocked", "#ff6b6b", f"Is email (<b>{em}</b>) ko block kar diya gaya hai. Mazeed maloomat ke liye rabta karein:")

elif status == "off":
    _show_card("🔴 Service Offline", "#ff6b6b", f"<b>{CONFIG['tool_name']}</b> filhal band hai. Update ke liye hamara WhatsApp channel join karein:")

else:
    _show_card("🔒 Access Denied", "#ff6b6b", f"Email <b>{em}</b> registered nahi hai. Access hasil karne ke liye rabta karein:")
```
