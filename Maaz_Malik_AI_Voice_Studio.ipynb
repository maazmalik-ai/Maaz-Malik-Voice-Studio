```python
#@title 🔐 Step 1 — Malik Voice Studio Access Check { display-mode: "form" }
EMAIL = '' #@param {type:"string"}

# ==========================================================
# 1. SETTINGS (Edit your details here anytime)
# ==========================================================
CONFIG = {
    "tool_name": "Malik Voice Studio",
    "owner_name": "Maaz",
    "admin_emails": ["maaz.malik0684@gmail.com"],
    "api_url": "https://script.google.com/macros/s/AKfycbwkB-r1cimQazOKJO2Cag6edob7njP6Y_-oOvM2mMA6TnZVM8eAJ7ExYe8NxRSGFmzQiQ/exec",
    "secret_key": "malik_voice_secret_2026",
    "support_email": "maaz.malik0684@gmail.com",
    "whatsapp_number": "",
    "whatsapp_link": "",
    "wa_channel_name": "",
    "wa_channel_link": "",
    "social_links": {
        "YouTube": "",
        "Instagram": "",
        "TikTok": "",
    },
}

# ==========================================================
# 2. ACCESS VERIFICATION ENGINE (Do not modify)
# ==========================================================
import urllib.request, urllib.parse, json, secrets, builtins
from IPython.display import display, HTML

builtins._ok = False
builtins._malik_session = None
builtins._malik_active_user = None

def _show_card(title, color, message):
    btns = []
    wa_link = (CONFIG.get("whatsapp_link") or "").strip()
    if wa_link and "PASTE_" not in wa_link and "YAHAN_" not in wa_link:
        btns.append(f"<a href='{wa_link}' target='_blank' style='background:#25d366;color:#fff;padding:9px 16px;border-radius:8px;text-decoration:none;font-weight:700;margin-right:8px;display:inline-block;margin-top:10px;'>💬 WhatsApp</a>")

    ch_link = (CONFIG.get("wa_channel_link") or "").strip()
    ch_name = (CONFIG.get("wa_channel_name") or "WhatsApp Channel").strip()
    if ch_link and "PASTE_" not in ch_link and "YAHAN_" not in ch_link:
        btns.append(f"<a href='{ch_link}' target='_blank' style='background:#128c7e;color:#fff;padding:9px 16px;border-radius:8px;text-decoration:none;font-weight:700;margin-right:8px;display:inline-block;margin-top:10px;'>📢 {ch_name}</a>")

    for name, url in (CONFIG.get("social_links") or {}).items():
        u = (url or "").strip()
        if u and "PASTE_" not in u and "YAHAN_" not in u:
            btns.append(f"<a href='{u}' target='_blank' style='background:#2b2538;color:#fff;border:1px solid #554b6e;padding:8px 14px;border-radius:8px;text-decoration:none;font-size:.88em;margin-right:8px;display:inline-block;margin-top:10px;'>🔗 {name}</a>")

    contact = []
    if CONFIG.get("support_email"):
        contact.append(f"📧 <b>Email:</b> {CONFIG['support_email']}")
    wa_num = (CONFIG.get("whatsapp_number") or "").strip()
    if wa_num and "PASTE_" not in wa_num and "YAHAN_" not in wa_num:
        contact.append(f"💬 <b>WhatsApp:</b> {wa_num}")

    display(HTML(
        f"<div style='font-family:Inter,Arial,sans-serif;max-width:580px;background:#181320;border:1.5px solid {color};border-radius:14px;padding:22px 26px;color:#eee;'>"
        f"<div style='font-size:1.25em;font-weight:800;color:{color};'>{title}</div>"
        f"<div style='margin-top:10px;line-height:1.65;font-size:.95em;'>{message}</div>"
        f"<div style='margin-top:12px;line-height:1.8;font-size:.92em;'>{'<br>'.join(contact)}</div>"
        f"<div>{''.join(btns)}</div>"
        f"<div style='margin-top:14px;font-size:.8em;color:#998;'>{CONFIG['tool_name']} • Managed by {CONFIG['owner_name']}</div>"
        f"</div>"
    ))
    raise SystemExit("Access verification stopped. Please check the message above.")

em = EMAIL.strip().lower()
if not em:
    _show_card("⚠️ Email Required", "#f5b041", "Please enter your registered Gmail address in the <b>EMAIL</b> field above and run this cell again.")

api_url = (CONFIG.get("api_url") or "").strip()
if not api_url or "PASTE_" in api_url or "YAHAN_" in api_url or not api_url.startswith("http"):
    _show_card("⚙️ Configuration Error", "#f5b041", "The API endpoint URL is missing or invalid. Please configure <b>api_url</b> in the settings.")

# Step A: Verify actual logged-in Google account
try:
    from google.colab import auth
    auth.authenticate_user()
    import google.auth, google.auth.transport.requests
    creds, _ = google.auth.default()
    creds.refresh(google.auth.transport.requests.Request())
    req = urllib.request.Request(f"https://www.googleapis.com/oauth2/v1/userinfo?access_token={creds.token}")
    colab_email = json.loads(urllib.request.urlopen(req, timeout=10).read().decode()).get("email", "").lower()
except Exception:
    colab_email = ""

if not colab_email:
    _show_card("⚠️ Google Login Failed", "#ff6b6b", "Unable to verify your Google account. Please sign in to Colab and try again.")

is_admin = colab_email in [a.strip().lower() for a in CONFIG.get("admin_emails", [])]
print(f"🔍 Logged-in Google Account: {colab_email}")

if (not is_admin) and (colab_email != em):
    _show_card("🚫 Account Mismatch", "#ff6b6b", f"Your logged-in Google account (<b>{colab_email}</b>) does not match the entered email (<b>{em}</b>). Please use your approved Google account.")

# Step B: Check Start, Expire, and Status from Google Sheet API
try:
    q_url = f"{api_url}?key={urllib.parse.quote(CONFIG['secret_key'])}&email={urllib.parse.quote(em)}"
    data = json.loads(urllib.request.urlopen(q_url, timeout=15).read().decode())
except Exception:
    _show_card("🌐 Connection Error", "#f5b041", "Could not connect to the verification server. Please check your internet connection and try again.")

status  = data.get("status", "")
plan    = data.get("plan", "Standard")
start_t = data.get("start", "")
exp_t   = data.get("expires", "")

if status == "ok":
    builtins._ok = True
    builtins._malik_session = secrets.token_hex(24)
    builtins._malik_active_user = {"email": em, "plan": plan, "tool_name": CONFIG["tool_name"]}
    print(f"✅ Access Granted to {CONFIG['tool_name']}! | Plan: {plan}")
    if start_t: print(f"   🟢 Starts:  {start_t}")
    if exp_t:   print(f"   🔴 Expires: {exp_t}")
    print("👉 You can now run Step 2 below.")

elif status == "not_started":
    _show_card("🕒 Plan Not Started Yet", "#f5b041", f"Your access plan (<b>{plan}</b>) has not started yet.<br><b>Start Time:</b> {start_t}<br><b>Expiry Time:</b> {exp_t}")

elif status == "expired":
    _show_card("⏳ Subscription Expired", "#ff6b6b", f"Your access to <b>{CONFIG['tool_name']}</b> has expired.<br><b>Expired On:</b> {exp_t}<br>Please contact us to renew your subscription:")

elif status == "blocked":
    _show_card("🛑 Account Blocked", "#ff6b6b", f"Access for <b>{em}</b> has been blocked on <b>{CONFIG['tool_name']}</b>. Please contact support for assistance:")

elif status == "off":
    _show_card("🔴 Service Offline", "#ff6b6b", f"<b>{CONFIG['tool_name']}</b> is temporarily offline. Please check our updates below:")

else:
    _show_card("🔒 Access Denied", "#ff6b6b", f"The email <b>{em}</b> is not approved for <b>{CONFIG['tool_name']}</b>. Please contact us to request access:")
```
